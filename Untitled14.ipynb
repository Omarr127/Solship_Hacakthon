```python
# [CELL 1]
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm

# 1. Load the Solship 2024 Data
solship_df = pd.read_csv('solship_training_data_2024.csv')
solship_df['timestamp'] = pd.to_datetime(solship_df['timestamp'])

# 2. Load the Open-Meteo Weather Data (Skip first 3 metadata rows)
weather_file = "open-meteo-46.15N9.89E308m (2).csv"
weather_df = pd.read_csv(weather_file, skiprows=3)

# Rename to match Solship data and convert to datetime
weather_df.rename(columns={'time': 'timestamp'}, inplace=True)
weather_df['timestamp'] = pd.to_datetime(weather_df['timestamp'])

print("Solship Data Shape:", solship_df.shape)
print("Weather Data Shape:", weather_df.shape)
```

```python
# [CELL 2]
# The weather data is hourly. We MUST resample it to 15-minute intervals
# to match the Solship data, using linear interpolation so there are no NaNs.

weather_df.set_index('timestamp', inplace=True)
weather_15min_df = weather_df.resample('15min').interpolate(method='linear')
weather_15min_df.reset_index(inplace=True)

# Merge datasets on the timestamp
merged_df = pd.merge(solship_df, weather_15min_df, on='timestamp', how='inner')

# Clean up messy weather column names
merged_df.columns = [
    col.lower().replace(' ', '_').replace('(', '').replace(')', '')
    .replace('°c', 'c').replace('%', 'pct').replace('m²', 'm2').replace('/', '_')
    for col in merged_df.columns
]

print("Merged Data Shape:", merged_df.shape)
```

```python
# [CELL 3]
# We extract time features FIRST so we can use them in our EDA plots.
merged_df['hour'] = merged_df['timestamp'].dt.hour
merged_df['day_of_week'] = merged_df['timestamp'].dt.dayofweek # 0=Mon, 6=Sun
merged_df['month'] = merged_df['timestamp'].dt.month
merged_df['is_weekend'] = merged_df['day_of_week'].isin([5, 6]).astype(int)

display(merged_df.head(3))
```

```python
# ==========================================
# CELL 3: What exactly drives the Electrical Load?
# ==========================================
plt.figure(figsize=(8, 8))

# Select only the features that could theoretically impact the load
features_to_check = [
    'load_kw', # Our target
    'temperature_2m_c',
    'relative_humidity_2m_pct',
    'cloud_cover_pct',
    'precipitation_mm',
    'wind_speed_10m_km_h',
    'hour',
    'month',
    'day_of_week'
]

# Calculate correlations specifically against 'load_kw' and sort them
corr_with_load = merged_df[features_to_check].corr()[['load_kw']].sort_values(by='load_kw', ascending=False)

# Plot a 1-column heatmap
sns.heatmap(corr_with_load, annot=True, cmap='coolwarm', fmt=".3f",
            vmin=-1, vmax=1, annot_kws={"size": 12}, linewidths=1, cbar_kws={'label': 'Correlation with Load'})

plt.title('Feature Correlation with Load (kW)', fontsize=16, pad=15)
plt.yticks(fontsize=12)
plt.xticks([]) # Hide the x-axis label since it just says 'load_kw'
plt.tight_layout()
plt.show()

print("-" * 80)
print("COMMENT: This 1D heatmap is laser-focused on our target variable: load_kw.")
print("Features at the very top (positive correlation) and the very bottom (negative correlation)")
print("are the most important ones to feed into your XGBoost/LightGBM model.")
print("Features hovering near 0.000 are practically useless and can be ignored.")
print("-" * 80)
```

```python
# ==========================================
# CELL 4: Daily Habits (Hour of Day)
# ==========================================
plt.figure(figsize=(14, 7))

sns.boxplot(data=merged_df, x='hour', y='load_kw', palette='viridis')
plt.title('Load Variance by Hour of Day', fontsize=16)
plt.xlabel('Hour of the Day (0 to 23)', fontsize=14)
plt.ylabel('Electrical Load (kW)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.show()

print("-" * 80)
print("COMMENT: This shows the daily routine. Notice how the boxes shift up and down.")
print("The height of the box shows how unpredictable the load is at that specific hour.")
print("Tall boxes mean the XGBoost model will need extra weather features to guess the load accurately there.")
print("-" * 80)
```

```python
# ==========================================
# CELL 6: Memory of the Past (Lags)
# ==========================================
plt.figure(figsize=(16, 6))

# We plot 100 lags (100 * 15 mins = 25 hours)
sm.graphics.tsa.plot_acf(merged_df['load_kw'].dropna(), lags=100, ax=plt.gca(), markersize=5)
plt.title('Autocorrelation of Electrical Load', fontsize=16)
plt.xlabel('Lag (1 step = 15 minutes. 96 steps = 24 hours)', fontsize=14)
plt.ylabel('Correlation Score', fontsize=14)
plt.grid(True, alpha=0.3)
plt.show()

print("-" * 80)
print("COMMENT: This is the most important plot for LightGBM/XGBoost feature engineering!")
print("Look for the massive spike at lag 96 (the 24-hour mark). This proves that the load right now")
print("is highly correlated with the load at this exact time yesterday. You MUST engineer a 24h lag feature.")
print("-" * 80)
```
